Sachant que L_b(p, t) = 1_{p≠t}, écrivons le code qui permet de déterminer le meilleur modèle entre g1 et g2 définis par leur matrice de confusion.

```python
import numpy as np
```

#### Initialisation des matrices de confusion de chaque modèle

```python
g1 = np.array([[44, 0, 0], [5, 62, 1], [1, 8, 54]])
g2 = np.array([[44, 4, 5], [2, 64, 3], [4, 2, 47]])
```

#### Risque empirique

```python
def risque_emp(matrice_confusion):
    tot = np.sum(matrice_confusion)
    nb_lignes = matrice_confusion.shape[0] # on récupère le nombre de lignes
    bonnes_pred = 0

    # on va soustraire la somme des éléments sur la diagonale à la somme totale des observations
    for i in range(nb_lignes):
        bonnes_pred += matrice_confusion[i, i]

    risque_emp = (tot - bonnes_pred) / tot
    return risque_emp
```

#### Comparaison pour trouver le modèle qui minimise le risque empirique

```python
risque_emp_g1 = risque_emp(g1)
risque_emp_g2 = risque_emp(g2)

print(f"Risque empirique pour g1 : {risque_emp_g1}")
print(f"Risque empirique pour g2 : {risque_emp_g2}")

if risque_emp_g1 < risque_emp_g2 :
    print("Le meilleur modèle est g1")
elif risque_emp_g2 < risque_emp_g1 :
    print("Le meilleur modèle est g2")
else:
    print("Il n'y a pas de meilleur modèle, les deux sont équivalents")
```
